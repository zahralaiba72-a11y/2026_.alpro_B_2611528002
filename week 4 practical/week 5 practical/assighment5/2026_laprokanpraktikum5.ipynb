## Basic theory
Definite Iteration (for Loop): Executes a block of code a predetermined number of times by iterating over items in a sequence (such as a string, list, or range() generator).

The range() Generator Function: Produces an immutable sequence of integers defined by range(start, stop[, step]). It operates up to, but does not include, the stop boundary (e.g., range(1, 5) yields 1, 2, 3, 4).   

Nested Loop Structures: Consists of one or more loops placed inside another loop. For every single iteration of the outer loop, the inner loop executes its entire cycle completely, enabling multi-dimensional sequence mapping and complex matrix rendering.   

Stream Output Formatting (end Parameter): Modifies the default newline behavior of Python’s print() function (e.g., print(..., end="")), enabling characters to be printed continuously on the same horizontal line during loop iterations.   

Loop Index Offsetting and Step Direction: Controls iteration dynamics by applying arithmetic expressions to loop variables (e.g., i + 1 for 1-based indexing) or negative step sizes (range(N, 0, -1)) for descending counts.   

Symmetric Pattern Logic and Grid Alignment: Combines nested loops with variable space counts and palindromic number progression to build geometrically aligned visual shapes along horizontal and vertical axes.

### 2.2.1 jumlah_genap_2611528002
Run the code cell below so that Jupyter automatically saves the code along with its output as a single unit (no separate screenshot needed for the execution output).

```python
#create a file with name jumlah_genap_2611538002.py
#create a program for loop in python
#variable names are appended with your student number
#this program uses the input() function
loop = int(input("Enter the limit value: "))
sum = 0
for i in range(1, loop + 1):
    if i % 2 == 0:
        print(i, end=" ")
        sum = sum + i
        if i < loop:
            print(" +", end=" ")
        else:
            print(" = ", sum, end=" ")
            print()
print("sum of even numbers =", sum)


```

### Output Explanation:

The user entered a limit value of 50 (loop = 50). The program sets up a for loop using range(1, 51) to iterate through all integers from 1 to 50 and filters for even numbers using the modulus condition (if i % 2 == 0).

Execution Path:

For each even integer i∈[2,4,6,…,50], the program prints the number i.

The running total sum accumulates each even number (sum = sum + i).

Since every even number i<50 satisfies i < loop, a + symbol is printed after each number except for the final even number (50).

When i=50, the else branch executes, printing = followed by the total accumulated sum (650) directly on the same line.

Finally, the last print("sum of even numbers =", sum) displays the final summary line.

## Tugas Praktikum

Students are asked to design an interactive console program that prints a two-dimensional geometric pattern in the form of a Palindromic Crystal Hourglass, neatly enclosed within a fenced border frame. Unlike the basic pattern, this program requires forming number sequences that move dynamically downward and upward palindromically on each row, aligned with the crystal axis at the center of the shape.
The program accepts a positive integer input N from the user, representing the scale size of the hourglass (the highest number at the top and the base).


**Kode Jawaban :**

a.  Kode Program "Alpro Tugas5_B_2611538002"

```python
# Program Framed Palindromic Crystal Hourglass (Week 5)
# File Name: tugas5_8002.py
# All variables end with the last 4 digits of the student ID (NIM)

print("=== FRAMED PALINDROMIC CRYSTAL HOURGLASS PROGRAM (WEEK 5) ===")
n_8002 = int(input("Enter hourglass scale size (N): "))

# 1. TOP HORIZONTAL BORDER
# Format: # + '=' repeated (4 * N + 5) times + #
print("#", end="")
for border_8002 in range(4 * n_8002 + 5):
    print("=", end="")
print("#")

# 2. PHASE 1: UPPER HOURGLASS (Row N down to 1)
for row_8002 in range(n_8002, 0, -1):
    print("| ", end="")  # Left vertical border + 1 padding space
    
    # Left balancing spaces: 2 * (N - row)
    for spasi_8002 in range(2 * (n_8002 - row_8002)):
        print(" ", end="")
        
    # Descending number sequence: row down to 1
    for num_8002 in range(row_8002, 0, -1):
        print(num_8002, end=" ")
        
    # Center crystal axis
    print("<*>", end="")
    
    # Ascending number sequence: 1 up to row
    for num_8002 in range(1, row_8002 + 1):
        print(f" {num_8002}", end="")
        
    # Right balancing spaces: 2 * (N - row)
    for spasi_8002 in range(2 * (n_8002 - row_8002)):
        print(" ", end="")
        
    print(" |")  # Right padding space + vertical border

# 3. PHASE 2: HOURGLASS CENTER AXIS (Singularity)
print("|", end="")
# Left balancing spaces: (2 * N + 1)
for spasi_8002 in range(2 * n_8002 + 1):
    print(" ", end="")

# Center crystal axis
print("<*>", end="")

# Right balancing spaces: (2 * N + 1)
for spasi_8002 in range(2 * n_8002 + 1):
    print(" ", end="")
print("|")

# 4. PHASE 3: LOWER HOURGLASS (Row 1 up to N)
for row_8002 in range(1, n_8002 + 1):
    print("| ", end="")  # Left vertical border + 1 padding space
    
    # Left balancing spaces: 2 * (N - row)
    for spasi_8002 in range(2 * (n_8002 - row_8002)):
        print(" ", end="")
        
    # Descending number sequence: row down to 1
    for num_8002 in range(row_8002, 0, -1):
        print(num_8002, end=" ")
        
    # Center crystal axis
    print("<*>", end="")
    
    # Ascending number sequence: 1 up to row
    for num_8002 in range(1, row_8002 + 1):
        print(f" {num_8002}", end="")
        
    # Right balancing spaces: 2 * (N - row)
    for spasi_8002 in range(2 * (n_8002 - row_8002)):
        print(" ", end="")
        
    print(" |")  # Right padding space + vertical border

# 5. BOTTOM HORIZONTAL BORDER
print("#", end="")
for border_8002 in range(4 * n_8002 + 5):
    print("=", end="")
print("#")
```

## OUTPUT EXPLANATION

Condition Check: The user entered an hourglass scale size of 12 (N = 12). The program computes the horizontal border length using the formula 4×N+5=53 = characters.   

Execution Path:

Horizontal Borders: The top and bottom borders are constructed using nested loops to print # at the corners with 53 = characters between them (#=====================================================#).   

Phase 1 (Upper Hourglass): The outer loop iterates row downward from 12 to 1. Each row prints:   

Left frame boundary | .   

2×(12−row) balancing spaces.   

Descending sequence from row down to 1.   

Center crystal axis <*>.  

Ascending sequence from 1 up to row.   

Right balancing spaces and vertical frame boundary |.   

Phase 2 (Center Axis / Singularity): The center row prints the vertical frame boundaries | with 2×12+1=25 padding spaces on both sides of the center crystal axis <*>.   

Phase 3 (Lower Hourglass): The outer loop iterates row upward from 1 to 12, recreating the palindromic number sequence expansion mirrored from Phase 1.
