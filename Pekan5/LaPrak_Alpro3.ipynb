## 2.1 Basic theory

Arithmetic Operators: Perform mathematical calculations including addition (+), subtraction (-), multiplication (*), floating-point division (/), integer division (//), modulo (%), and exponentiation (**).   

Assignment Operators: Store values into variables. Augmented assignments combine arithmetic and assignment operations into single operations such as +=, -=, *=, /=, //=, %=, and **=.   

Comparison Operators: Evaluate relational equality between two operands (>, <, >=, <=, ==, !=), returning a boolean result (True or False).   

Logical Operators: Combine boolean expressions using conjunction (and), disjunction (or), and negation (not).   

Membership Operators: Determine element presence within sequence objects using in and not in.   

Identity Operators: Verify whether two variables share identical memory locations using is and is not, contrasting with value equality (==).   

Bitwise Operators: Perform bit-level operations on integer binary representations using AND (&), OR (|), XOR (^), NOT (~), left-shift (<<), and right-shift (>>).

```python
# Create a file named bitwise_NIM.py
# Variable names are appended with the last 4 digits of the student ID (example: angka1_1234)
# This program uses the input() function

print("\n==============================")
print("3. BITWISE OPERATOR")
print("==============================")

angka1 = int(input("Enter bitwise number-1: "))
angka2 = int(input("Enter bitwise number-2: "))

print("\nNumbers in decimal and binary form")
print("angka1 =", angka1, "| binary =", bin(angka1))
print("angka2 =", angka2, "| binary =", bin(angka2))

# Bitwise AND
hasil = angka1 & angka2
print("\nBitwise AND (&)")
print(angka1, "&", angka2, "=", hasil)
print("Binary result =", bin(hasil))
print("Binary result (8 bit) =", format(hasil, "08b"))

# Bitwise OR
hasil = angka1 | angka2
print("\nBitwise OR (|)")
print(angka1, "|", angka2, "=", hasil)
print("Binary result =", bin(hasil))
print("Binary result (8 bit) =", format(hasil, "08b"))

# Bitwise XOR
hasil = angka1 ^ angka2
print("\nBitwise XOR (^)")
print(angka1, "^", angka2, "=", hasil)
print("Binary result =", bin(hasil))
print("Binary result (8 bit) =", format(hasil, "08b"))

# Bitwise NOT
hasil = ~angka1
print("\nBitwise NOT (~)")
print("~", angka1, "=", hasil)
print("Binary result =", bin(hasil))
print("Binary result (8 bit) =", format(hasil, "08b"))

# Bitwise left shift
jumlah_geser = int(input("\nEnter the number of bit shifts: "))

hasil = angka1 << jumlah_geser
print("\nBitwise left shift (<<)")
print(angka1, "<<", jumlah_geser, "=", hasil)
print("Binary result =", bin(hasil))
print("Binary result (8 bit) =", format(hasil, "08b"))

# Bitwise right shift
hasil = angka1 >> jumlah_geser
print("\nBitwise right shift (>>)")
print(angka1, ">>", jumlah_geser, "=", hasil)
print("Binary result =", bin(hasil))
print("Binary result (8 bit) =", format(hasil, "08b"))
```

When executing the program with inputs `angka1 = 33`, `angka2 = 24`, and shift count `11`:
1. Binary Representation: The program converts `33` to binary `00100001` and `24` to `00011000`.
2. Bitwise AND (`&`): Compares corresponding bits and sets to `1` only if both are `1`. Since no positions overlap, the result is `0` (`00000000`).
3. Bitwise OR (`|`): Combines active bits from both numbers, resulting in `00111001` (`57` in decimal).
4. Bitwise XOR (`^`): Sets bits to `1` where values differ. With non-overlapping inputs, this matches the OR operation, returning `57`.
5. Bitwise NOT (`~`): Negates all bits using standard two's complement syntax `~x = -(x + 1)`, turning `33` into `-34`.
6. Bitwise Left Shift (`<<`): Moves bits 11 positions to the left (multiplying by $2^{11}$), resulting in `67584`.
7. Bitwise Right Shift (`>>`): Moves bits 11 positions to the right (performing floor division by $2^{11}$), reducing the value to `0`.

2. Membership `not in`: Evaluates `4 not in [2, 3, 6, 12, 32, 11]` and returns `True` confirming the number is absent.
3. Identity `is`: Compares `objek1` and `objek2`. Because `objek2` points directly to the memory reference of `objek1`, `objek1 is objek2` yields `True`.
4. Identity `is not`: Compares `objek1` and `objek3`. Because `objek3` was instantiated as a shallow copy (`data.copy()`), it occupies a separate memory location, so `objek1 is not objek3` returns `True`.
5. Identity vs Value Comparison: 
   - `objek1 is objek3` evaluates to `False` because identity requires matching memory addresses.
   - `objek1 == objek3` evaluates to `True` because standard equality only verifies that the list contents and sequence are identical.

. Modulo (%): Identifies leftover integer remainders when total spend is divided by quantity units.   

Q2. Distinction Between Logical and, or, and not

. and: Returns True only if all expressions evaluate to True (e.g., verifying is_member AND total spend ≥Rp200,000).   

. or: Returns True if at least one sub-condition evaluates to True (e.g., qualifying for promos via spending OR buying ≥3 items).   

. not: Inverts boolean results, turning True to False and vice versa.   

Q3. Purpose of Membership Operators (in / not in)

. in: Scans the promo list ["HEMAT10", "HEMAT20", "GRATISONGKIR"] and yields True if the user's promo code matches any list item.   

. not in: Returns True if the input string is absent from the valid list.   

Q4. Identity (is) vs Equality (==) Comparison

. == (Value Equality): Evaluates whether two objects share identical data values regardless of memory storage address.   

. is (Identity Comparison): Verifies whether two references point directly to the exact same memory address. Separate objects containing identical elements yield True under == but False under is.

 Bitwise operators compress complex state flags into binary positions (0001 Member, 0010 Spend, 0100 Quantity, 1000 Promo):   

. Bitwise OR (|): Combines binary flags into an integrated bitmask access integer.   

. Bitwise AND (&): Isolates specific bit positions to evaluate specific privileges.   

. Bitwise XOR (^): Identifies structural differences between actual status bitmasks and baseline permissions.
