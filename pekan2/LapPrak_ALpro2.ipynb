## 2.1 Dasar Teori/Basic theory

Python Programming Language: Python is a high-level, interpreted programming language with a clean and intuitive syntax focused on code readability, making it ideal for learning fundamental programming logic without complex structural overhead.

Input and Output (I/O) Functions: The input() function is used to capture text data from the user through the console, while the print() function displays variables, strings, and execution results to the screen.

Data Types and Type Casting: Python supports primitive and non-primitive data types including strings (str), integers (int), floats (float), complex numbers (complex), and booleans (bool). Because input() defaults to string data, explicit casting (int(), float()) is necessary to convert values for numerical operations.

Constants and Variable Naming: Constants represent fixed values that remain unchanged during execution (e.g., PASSING_SCORE_8002 = 75.0). Variable names follow case-sensitive rules and include the student ID suffix (_8002) for proper tracking.

Relational Operators and Boolean Logic: Relational operators (such as >=) evaluate conditions to return boolean values (True or False), enabling automated decision-making and status verification within the program.
